```python
from dataset_summarizer import DatasetSummarizer
from hf_loader import HFDatasetLoader
```

```python
# Puedes elegir cualquiera de las 11 tablas (ej: 'match_stats', 'players', 'teams', 'leagues', etc.)
TABLA_ELEGIDA = "match_stats"

loader = HFDatasetLoader(
    dataset_name="eatpizzanot/soccer-dataset",
    name=TABLA_ELEGIDA,
    split="train",
)
df_original = loader.load_data()

print(f"--- DataFrame Cargado Original ('{TABLA_ELEGIDA}') ---")
print(f"Filas totales: {len(df_original)}")
display(df_original.head())
```

```python
# Transformación mínima: Filtrar filas eliminando nulos de la primera columna real
primera_columna = df_original.columns[0]

# Crear el DataFrame transformado mediante un filtrado
df_transformado = df_original.dropna(subset=[primera_columna]).copy()

print(f"--- DataFrame Transformado (Sin nulos en '{primera_columna}') ---")
print(f"Filas resultantes: {len(df_transformado)}")
display(df_transformado.head())
```

```python
# Pasar el DataFrame transformado a la segunda clase
summarizer = DatasetSummarizer(
    df=df_transformado, dataset_name=f"Soccer Dataset - {TABLA_ELEGIDA}"
)

# Generar el archivo .txt
archivo_generado = summarizer.summarize_to_txt(
    filename=f"resumen_{TABLA_ELEGIDA}_transformado"
)
```

Explicación de la Transformación:
1. Se seleccionó la columna principal del dataset ('{}').
2. Se aplicó un filtro de limpieza con `.dropna(subset=[...])` para eliminar las filas con valores nulos en esa columna.
3. El número de registros se redujo de {:,} a {:,} filas.
4. El resumen '.txt' resultante refleja exactamente las dimensiones y estadísticas del DataFrame limpio.
