# EA1: carga, transformaci?n y resumen de datos

Este cuaderno utiliza las clases del proyecto para cargar `match_stats`, aplicar cuatro transformaciones sencillas y generar un informe que corresponde al DataFrame final.

```python
from dataset_summarizer import DatasetSummarizer
from hf_loader import HFDatasetLoader
```

## 1. Cargar datos desde Hugging Face

La clase de carga obtiene el conjunto y lo convierte en un DataFrame de Pandas.

```python
nombre_conjunto = "eatpizzanot/soccer-dataset"
nombre_tabla = "match_stats"

cargador = HFDatasetLoader(
    dataset_name=nombre_conjunto,
    name=nombre_tabla,
    split="train",
)
df_original = cargador.load_data()

print(f"Dimensiones de los datos originales: {df_original.shape}")
display(df_original.head())
```

## 2. Aplicar cuatro transformaciones

1. Renombrar las columnas que se usar?n.
2. Seleccionar solo las estad?sticas relevantes.
3. Quitar filas sin identificador o sin datos de tiros totales.
4. Calcular el total de tiros de ambos equipos.

```python
df_transformado = df_original.rename(
    columns={
        "fixture_id": "identificador_partido",
        "home_shots_total": "tiros_totales_local",
        "away_shots_total": "tiros_totales_visitante",
        "home_shots_on_goal": "tiros_a_puerta_local",
        "away_shots_on_goal": "tiros_a_puerta_visitante",
        "home_possession": "posesion_local",
        "away_possession": "posesion_visitante",
        "home_fouls": "faltas_local",
        "away_fouls": "faltas_visitante",
    }
)
print(f"1. Columnas renombradas: {df_transformado.shape}")
display(df_transformado.head())
```

```python
columnas = [
    "identificador_partido",
    "tiros_totales_local",
    "tiros_totales_visitante",
    "tiros_a_puerta_local",
    "tiros_a_puerta_visitante",
    "posesion_local",
    "posesion_visitante",
    "faltas_local",
    "faltas_visitante",
]
df_transformado = df_transformado[columnas].copy()
print(f"2. Columnas seleccionadas: {df_transformado.shape}")
display(df_transformado.head())
```

```python
filas_antes = len(df_transformado)
df_transformado = df_transformado.dropna(
    subset=["identificador_partido", "tiros_totales_local", "tiros_totales_visitante"]
).copy()
print(
    "3. Filas completas para el an?lisis: "
    f"{filas_antes} ? {len(df_transformado)}"
)
display(df_transformado.head())
```

```python
df_transformado["tiros_totales_partido"] = (
    df_transformado["tiros_totales_local"]
    + df_transformado["tiros_totales_visitante"]
)
print(f"4. Total de tiros calculado: {df_transformado.shape}")
display(df_transformado.head())
```

## 3. Generar el resumen del DataFrame transformado

El archivo de texto se crea a partir del mismo `df_transformado` que se muestra arriba. Contiene sus dimensiones, columnas, tipos de datos, valores faltantes, estad?sticas y una vista previa.

```python
resumidor = DatasetSummarizer(
    df=df_transformado,
    dataset_name="F?tbol - estad?sticas de partidos transformadas",
)
archivo_resumen = resumidor.summarize_to_txt(
    output_folder=".",
    filename="resumen_match_stats_transformado",
)
print(f"El resumen de esta transformaci?n se guard? en: {archivo_resumen}")
print(f"Dimensiones resumidas: {df_transformado.shape}")
```
